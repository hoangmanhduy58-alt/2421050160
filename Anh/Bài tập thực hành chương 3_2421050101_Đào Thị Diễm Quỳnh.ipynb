**Bài 10: Tính điểm học tập của bạn**

```python
diem_he_10 = [8.4, 6.5, 7.3, 2.6, 9.0, 5.8, 6.0, 9.7, 8.1]

def quy_doi_diem(score):
    if score >= 9.0:
        return 4.0, 'A+'
    elif score >= 8.5:
        return 3.7, 'A'
    elif score >= 8.0:
        return 3.5, 'B+'
    elif score >= 7.0:
        return 3.0, 'B'
    elif score >= 6.5:
        return 2.5, 'C+'
    elif score >= 5.5:
        return 2.0, 'C'
    elif score >= 5.0:
        return 1.5, 'D+'
    elif score >= 4.0:
        return 1.0, 'D'
    else:
        return 0.0, 'F'

#Yêu cầu 1: Tạo một danh sách các điểm chữ và điểm hệ 4 tương ứng
diem_he_4 = []
diem_chu = []

for score in diem_he_10:
    gpa4, letter = quy_doi_diem(score)
    diem_he_4.append(gpa4)
    diem_chu.append(letter)

#Yêu cầu 2: Tính điểm trung bình Hệ 10 và Hệ 4
dtb_he_10 = sum(diem_he_10) / len(diem_he_10)
dtb_he_4 = sum(diem_he_4) / len(diem_he_4)

print(f"Điểm hệ 10:        {diem_he_10}")
print(f"Điểm chữ tương ứng: {diem_chu}")
print("-------Điểm Trung Bình-------")
print(f"Tổng số môn học: {len(diem_he_10)}")
print(f"ĐTB hệ 10:  {dtb_he_10}")
print(f"ĐTB hệ 4:   {dtb_he_4}")
```

**Bài 11: Kiểm tra N có là số nguyên tố?**

```python
import math

def is_prime(n):
    if n <= 1:
        return False
    for i in range(2, int(math.isqrt(n)) + 1):
        if n % i == 0:
            return False
    return True

def main():
    try:
        n = int(input("Nhập vào một số nguyên dương N (N>1): "))
        if is_prime(n):
            print(f"Số  {n}  là số nguyên tố!")
        else:
            print(f"Số  {n}  không phải là số nguyên tố!")
    except ValueError:
        print("Vui lòng nhập một số nguyên hợp lệ!")

if __name__ == "__main__":
    main()
```

**Bài 12: Dãy số nguyên tố**

```python
import math

def is_prime(num):
    if num < 2:
        return False
    for i in range(2, int(math.isqrt(num)) + 1):
        if num % i == 0:
            return False
    return True

def main():
    try:
        n = int(input("Nhập vào một số nguyên dương N (N>1): "))
        if n > 1:
            prime_list = []
            for i in range(2, n + 1):
                if is_prime(i):
                    prime_list.append(str(i))

            print(f"Các số nguyên tố từ 1 tới  {n} :")
            print(", ".join(prime_list))
        else:
            print("Vui lòng nhập N > 1!")
    except ValueError:
        print("Vui lòng nhập số nguyên hợp lệ!")

if __name__ == "__main__":
    main()
```

**Bài 13: Đổi số từ thập phân sang nhị phân**

```python
def print_division_steps(n):
    print(f"\n{'Dòng':<6}{'Phần dư':<10}{'Thương':<10}{'Chia cho'}")
    print("-" * 35)

    line = 1
    temp = n
    binary_bits = []

    while temp > 0:
        remainder = temp % 2
        next_temp = temp // 2
        binary_bits.append(str(remainder))

        print(f"{line:<6}{remainder:<10}{next_temp:<10}2")

        temp = next_temp
        line += 1

    binary_bits.reverse()
    return "".join(binary_bits)

def main():
    try:
        n = int(input("Nhập vào một số tự nhiên (>0): "))
        if n > 0:
            binary_result = print_division_steps(n)
            print("-" * 35)
            print(f"{n} (hệ 10) = {binary_result} (hệ 2)\n")
        else:
            print("Vui lòng nhập số tự nhiên lớn hơn 0!")
    except ValueError:
        print("Vui lòng nhập số nguyên hợp lệ!")

if __name__ == "__main__":
    main()
```

```python
#Yêu cầu 1: Hàm greeting()
from datetime import datetime

def greeting(ho_ten, nam_sinh):
    nam_hien_tai = datetime.now().year # Lấy năm hiện tại
    tuoi = nam_hien_tai - nam_sinh # Tính tuổi
    ten_in_hoa = ho_ten.upper()

    return f'Bạn "{ten_in_hoa}" năm nay {tuoi} tuổi!'
    
if __name__ == "__main__":
    name = input("Nhập vào họ tên: ")
    birth_year = int(input("Nhập vào năm sinh: "))
    thong_bao = greeting(name, birth_year)
    print(thong_bao)
```

```python
#Yêu cầu 4: Hàm bmi_show()
def bmi_show(chieu_cao, can_nang):
    bmi = can_nang / (chieu_cao ** 2)
 
    if bmi < 18.5:
        nhan_xet = "Underweight"
    elif 18.5 <= bmi <= 24.9:
        nhan_xet = "Normal Weight"
    elif 25.0 <= bmi <= 29.9:
        nhan_xet = "Overweight"
    else:
        nhan_xet = "Obese"

    return nhan_xet

if __name__ == "__main__":
    height = float(input("Nhập chiều cao (m): "))
    weight = float(input("Nhập cân nặng (kg): "))

    result = bmi_show(height, weight)
    print(f"Chỉ số BMI của bạn là: {bmi:.2f}")
    print(f"Nhận xét về cơ thể: {result}")
```

```python
#Yêu cầu 5: Hàm cal_point()
def quy_doi_he_4(diem10):

    if diem10 >= 9.0:
        return 4.0
    elif diem10 >= 8.5:
        return 3.7
    elif diem10 >= 8.0:
        return 3.5
    elif diem10 >= 7.0:
        return 3.0
    elif diem10 >= 6.5:
        return 2.5
    elif diem10 >= 5.5:
        return 2.0
    elif diem10 >= 5.0:
        return 1.5
    elif diem10 >= 4.0:
        return 1.0
    else:
        return 0.0

def cal_point(danh_sach_diem_10):
    if not danh_sach_diem_10:
        return 0.0, 0.0

    danh_sach_diem_4 = [quy_doi_he_4(d) for d in danh_sach_diem_10]

    dtb_he_10 = sum(danh_sach_diem_10) / len(danh_sach_diem_10)
    dtb_he_4 = sum(danh_sach_diem_4) / len(danh_sach_diem_4)

    return dtb_he_10, dtb_he_4

if __name__ == "__main__":
    diem_he_10 = [8.4, 6.5, 7.3, 2.6, 9.0, 5.8, 6.0, 9.7, 8.1]

    dtb10, dtb4 = cal_point(diem_he_10)

    print(f"Điểm hệ 10: {diem_he_10}")
    print(f"ĐTB hệ 10: {dtb10}")
    print(f"ĐTB hệ 4:  {dtb4}")
```

```python
#Yêu cầu 6: Hàm list_prime()
import math

def check_prime(num):

    if num < 2:
        return False
    for i in range(2, int(math.isqrt(num)) + 1):
        if num % i == 0:
            return False
    return True

def list_prime(n):
    """Trả về danh sách các số nguyên tố từ 1 đến n"""
    danh_sach_snt = [i for i in range(1, n + 1) if check_prime(i)]
    return danh_sach_snt

if __name__ == "__main__":
    n = int(input("Nhập vào số nguyên dương N (N > 1): "))

    ket_qua = list_prime(n)

    print(f"Các số nguyên tố từ 1 tới {n}:")
    print(ket_qua)
```

**Bài 15:**

```python
#Yêu cầu 1: Xây dựng lớp Person:
from datetime import datetime

class Person:
  
    def __init__(self, name='Nguyễn Văn A', year=2000, height=1.70, weight=65):
        self.name = name
        self.year = year
        self.height = height
        self.weight = weight

    def Geeting(self):
        current_year = datetime.now().year
        age = current_year - self.year
        print(f"My name is  {self.name}")
        print(f"I am  {age}  years old. Nice to meet you!")

    def BMI(self):
        bmi_value = self.weight / (self.height ** 2)
        return round(bmi_value, 1)

p1 = Person('Đào Thị Diễm Quỳnh', 2006, 1.52, 48.5)

p1.Geeting()

print('Chỉ số BMI:', p1.BMI())
```

```python
#Yêu cầu 2: Xây dựng lớp Vietnam, kế thừa từ lớp Person
class Vietnam(Person):
    def __init__(self, name='Nguyễn Văn A', year=2000, height=1.70, weight=65, nation='Kinh'):
      
        super().__init__(name, year, height, weight)
        
        self.nation = nation

   
    def Geeting(self):
        print(f"Tôi tên là:  {self.name} - Dân tộc: {self.nation}")
        print(f"Tôi sinh năm  {self.year} . Rất vui được làm quen với bạn!")

vn1 = Vietnam('Đào Thị Diễm Quỳnh', 2006, 1.52, 48.5, 'Kinh')

vn1.Geeting()

print('-----------------------------------------')
print('Chỉ số BMI:', vn1.BMI())
```
